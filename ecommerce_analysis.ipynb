```python
# SQL Query 1 - Total Revenue
query1 = '''
SELECT 
    ROUND(SUM(p.price * o.quantity), 2) as total_revenue,
    COUNT(o.order_id) as total_orders,
    COUNT(DISTINCT o.customer_id) as unique_customers
FROM orders o
JOIN products p ON o.product_id = p.product_id
WHERE o.status = 'Delivered'
'''
result1 = pd.read_sql_query(query1, conn)
print("=== BUSINESS OVERVIEW ===")
print(result1.to_string(index=False))
```

```python
# SQL Query 2 - Revenue by Category
query2 = '''
SELECT 
    p.category,
    COUNT(o.order_id) as total_orders,
    SUM(o.quantity) as total_units_sold,
    ROUND(SUM(p.price * o.quantity), 2) as revenue
FROM orders o
JOIN products p ON o.product_id = p.product_id
WHERE o.status = 'Delivered'
GROUP BY p.category
ORDER BY revenue DESC
'''
result2 = pd.read_sql_query(query2, conn)
print("=== REVENUE BY CATEGORY ===")
print(result2.to_string(index=False))

# Chart
plt.figure(figsize=(10,6))
sns.barplot(x='revenue', y='category', data=result2, palette='husl')
plt.title('Revenue by Product Category', fontsize=16)
plt.xlabel('Revenue (₹)')
plt.ylabel('Category')
plt.tight_layout()
plt.show()
```

```python
# SQL Query 3 - Top 5 Best Selling Products
query3 = '''
SELECT 
    p.product_name,
    p.category,
    p.price,
    SUM(o.quantity) as units_sold,
    ROUND(SUM(p.price * o.quantity), 2) as revenue
FROM orders o
JOIN products p ON o.product_id = p.product_id
WHERE o.status = 'Delivered'
GROUP BY p.product_id
ORDER BY revenue DESC
LIMIT 5
'''
result3 = pd.read_sql_query(query3, conn)
print("=== TOP 5 BEST SELLING PRODUCTS ===")
print(result3.to_string(index=False))

plt.figure(figsize=(10,6))
sns.barplot(x='revenue', y='product_name', data=result3, palette='viridis')
plt.title('Top 5 Products by Revenue', fontsize=16)
plt.xlabel('Revenue (₹)')
plt.ylabel('Product')
plt.tight_layout()
plt.show()
```

```python
# SQL Query 4 - Top Customers by Spending
query4 = '''
SELECT 
    c.customer_name,
    c.city,
    c.state,
    COUNT(o.order_id) as total_orders,
    ROUND(SUM(p.price * o.quantity), 2) as total_spent
FROM orders o
JOIN customers c ON o.customer_id = c.customer_id
JOIN products p ON o.product_id = p.product_id
WHERE o.status = 'Delivered'
GROUP BY c.customer_id
ORDER BY total_spent DESC
LIMIT 10
'''
result4 = pd.read_sql_query(query4, conn)
print("=== TOP 10 CUSTOMERS BY SPENDING ===")
print(result4.to_string(index=False))

plt.figure(figsize=(10,6))
sns.barplot(x='total_spent', y='customer_name', data=result4, palette='magma')
plt.title('Top 10 Customers by Total Spending', fontsize=16)
plt.xlabel('Total Spent (₹)')
plt.ylabel('Customer')
plt.tight_layout()
plt.show()
```

```python
# SQL Query 4 - Top Customers by Spending
query4 = '''
SELECT 
    c.customer_name,
    c.city,
    c.state,
    COUNT(o.order_id) as total_orders,
    ROUND(SUM(p.price * o.quantity), 2) as total_spent
FROM orders o
JOIN customers c ON o.customer_id = c.customer_id
JOIN products p ON o.product_id = p.product_id
WHERE o.status = 'Delivered'
GROUP BY c.customer_id
ORDER BY total_spent DESC
LIMIT 10
'''
result4 = pd.read_sql_query(query4, conn)
print("=== TOP 10 CUSTOMERS BY SPENDING ===")
print(result4.to_string(index=False))

plt.figure(figsize=(10,6))
sns.barplot(x='total_spent', y='customer_name', data=result4, palette='magma')
plt.title('Top 10 Customers by Total Spending', fontsize=16)
plt.xlabel('Total Spent (₹)')
plt.ylabel('Customer')
plt.tight_layout()
plt.show()
```

```python
# SQL Query 5 - Monthly Revenue Trend
query5 = '''
SELECT 
    strftime('%Y-%m', order_date) as month,
    COUNT(o.order_id) as orders,
    ROUND(SUM(p.price * o.quantity), 2) as revenue
FROM orders o
JOIN products p ON o.product_id = p.product_id
WHERE o.status = 'Delivered'
GROUP BY month
ORDER BY month
'''
result5 = pd.read_sql_query(query5, conn)
print("=== MONTHLY REVENUE TREND ===")
print(result5.to_string(index=False))

plt.figure(figsize=(12,6))
plt.plot(result5['month'], result5['revenue'], marker='o', linewidth=2, color='#e74c3c')
plt.fill_between(range(len(result5)), result5['revenue'], alpha=0.3, color='#e74c3c')
plt.xticks(range(len(result5)), result5['month'], rotation=45)
plt.title('Monthly Revenue Trend (2024)', fontsize=16)
plt.xlabel('Month')
plt.ylabel('Revenue (₹)')
plt.tight_layout()
plt.show()
```

```python
# SQL Query 6 - Order Status Analysis
query6 = '''
SELECT 
    status,
    COUNT(order_id) as count,
    ROUND(COUNT(order_id) * 100.0 / (SELECT COUNT(*) FROM orders), 2) as percentage
FROM orders
GROUP BY status
ORDER BY count DESC
'''
result6 = pd.read_sql_query(query6, conn)
print("=== ORDER STATUS BREAKDOWN ===")
print(result6.to_string(index=False))

plt.figure(figsize=(7,7))
plt.pie(result6['count'], labels=result6['status'],
        autopct='%1.1f%%', colors=['#2ecc71','#e74c3c','#f39c12'])
plt.title('Order Status Distribution', fontsize=16)
plt.show()
```

```python
# SQL Query 7 - Revenue by City
query7 = '''
SELECT 
    c.city,
    c.state,
    COUNT(o.order_id) as orders,
    ROUND(SUM(p.price * o.quantity), 2) as revenue
FROM orders o
JOIN customers c ON o.customer_id = c.customer_id
JOIN products p ON o.product_id = p.product_id
WHERE o.status = 'Delivered'
GROUP BY c.city
ORDER BY revenue DESC
'''
result7 = pd.read_sql_query(query7, conn)
print("=== REVENUE BY CITY ===")
print(result7.to_string(index=False))

plt.figure(figsize=(10,6))
sns.barplot(x='revenue', y='city', data=result7, palette='coolwarm')
plt.title('Revenue by City', fontsize=16)
plt.xlabel('Revenue (₹)')
plt.ylabel('City')
plt.tight_layout()
plt.show()
```

```python
# SQL Query 8 - Gender based buying behaviour
query8 = '''
SELECT 
    c.gender,
    COUNT(o.order_id) as total_orders,
    ROUND(SUM(p.price * o.quantity), 2) as total_revenue,
    ROUND(AVG(p.price * o.quantity), 2) as avg_order_value
FROM orders o
JOIN customers c ON o.customer_id = c.customer_id
JOIN products p ON o.product_id = p.product_id
WHERE o.status = 'Delivered'
GROUP BY c.gender
'''
result8 = pd.read_sql_query(query8, conn)
print("=== GENDER BASED BUYING BEHAVIOUR ===")
print(result8.to_string(index=False))
```

```python
# SQL Query 9 - Window Function - Customer Rank by Spending
query9 = '''
SELECT 
    c.customer_name,
    c.city,
    ROUND(SUM(p.price * o.quantity), 2) as total_spent,
    RANK() OVER (ORDER BY SUM(p.price * o.quantity) DESC) as spending_rank
FROM orders o
JOIN customers c ON o.customer_id = c.customer_id
JOIN products p ON o.product_id = p.product_id
WHERE o.status = 'Delivered'
GROUP BY c.customer_id
'''
result9 = pd.read_sql_query(query9, conn)
print("=== CUSTOMER RANKING BY SPENDING (Window Function) ===")
print(result9.to_string(index=False))
```

```python
# SQL Query 10 - CTE - High Value Customers above average spending
query10 = '''
WITH customer_spending AS (
    SELECT 
        c.customer_name,
        c.city,
        ROUND(SUM(p.price * o.quantity), 2) as total_spent
    FROM orders o
    JOIN customers c ON o.customer_id = c.customer_id
    JOIN products p ON o.product_id = p.product_id
    WHERE o.status = 'Delivered'
    GROUP BY c.customer_id
),
avg_spending AS (
    SELECT ROUND(AVG(total_spent), 2) as avg_spend 
    FROM customer_spending
)
SELECT 
    cs.customer_name,
    cs.city,
    cs.total_spent,
    av.avg_spend,
    CASE 
        WHEN cs.total_spent > av.avg_spend THEN 'High Value'
        ELSE 'Regular'
    END as customer_segment
FROM customer_spending cs, avg_spending av
ORDER BY cs.total_spent DESC
'''
result10 = pd.read_sql_query(query10, conn)
print("=== HIGH VALUE CUSTOMERS (CTE Query) ===")
print(result10.to_string(index=False))

conn.close()
print("\nDatabase connection closed.")
```
